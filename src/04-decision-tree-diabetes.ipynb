# ML Diabetes Tree

## Paso 1: Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
import pickle
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
```

## Paso 2: Lectura del conjunto

```python
df = pd.read_csv('../data/processed/diabetes_eda.csv')
df
```

## Paso 3: Split

```python
X = df.drop(['Outcome'], axis=1)
y = df['Outcome']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=18)
X_train.shape
```

## Paso 4: Modelado y ajuste 

```python
model = DecisionTreeClassifier(random_state=18)
model.fit(X_train, y_train)
```

## Paso 5: Análisis de variables

```python
plt.figure(figsize=(8, 8))

tree.plot_tree(model,
               feature_names=list(X_train.columns),
               filled=True)
plt.show()
# TODO: definir class_names. Ex: class_names=["iris setosa", "iris versicolor", "iris virginica"],
plt.show()
```

## Paso 6: Predicción

```python
y_pred = model.predict(X_test)
y_pred
```

## Paso 7: Métricas

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred)
```
